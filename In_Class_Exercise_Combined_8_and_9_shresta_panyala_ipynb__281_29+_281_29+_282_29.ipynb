### Q1. Load and inspect the dataset (2 points)
Show:

```python
# Read the uploaded CSV file
import pandas as pd

df = pd.read_csv(filename)

# 1. First 5 rows
print("First 5 rows:")
display(df.head())

# 2. Shape
print("Shape of dataset:", df.shape)

# 3. Column names
print("Column names:")
print(df.columns.tolist())
```

### Q2. Apply VADER to compute sentiment scores (3 points)
- Import VADER
- Download the lexicon if needed
- Create a new column called `compound`
- Show the first 5 rows of `Text` and `compound`

```python
!pip -q install nltk
import nltk
nltk.download('vader_lexicon')
from nltk.sentiment import SentimentIntensityAnalyzer
sia = SentimentIntensityAnalyzer()
```

```python
# Create compound sentiment score using VADER
df['compound'] = df['Text'].astype(str).apply(lambda x: sia.polarity_scores(x)['compound'])

# Show first 5 rows of Text and compound
df[['Text', 'compound']].head()
```

### Q3. Create a sentiment label from the compound score (3 points)
Create a new column called `PredictedSentiment` using this rule:
- `compound >= 0.05` → `positive`
- `compound <= -0.05` → `negative`
- otherwise → `neutral`

Then show the first 10 rows of `Text`, `compound`, and `PredictedSentiment`.

```python
# Create PredictedSentiment column based on compound score
def get_sentiment_label(compound):
    if compound >= 0.05:
        return 'positive'
    elif compound <= -0.05:
        return 'negative'
    else:
        return 'neutral'

df['PredictedSentiment'] = df['compound'].apply(get_sentiment_label)

# Show first 10 rows
df[['Text', 'compound', 'PredictedSentiment']].head(10)
```

### Q4. Compare review score and sentiment (2 points)
Group by `Score` and compute the **average compound score** for each score value.

```python
result = df.groupby('Score')['compound'].mean().reset_index()
result.columns = ['Score', 'Avg_Compound_Sentiment']
result
```

Your prompt should include:
- a clear instruction
- the three labels
- a request for a clean output format

Store your prompt in a Python variable called `zero_shot_prompt`, then print it.

```python
zero_shot_prompt = """
You are given a set of customer reviews. Your task is to determine the sentiment expressed in each review.

Possible sentiment labels:
- Positive
- Negative
- Neutral

Guidelines:
- Choose only one label per review
- Base your decision on the overall tone of the sentence

Output requirements:
Provide only the sentiment labels in a numbered list format as shown below:
1. label
2. label
3. label

Reviews to analyze:
1. The product arrived late and the quality was poor.
2. Amazing battery life and elegant design.
3. It is fine for daily use, nothing special.
"""

print(zero_shot_prompt)
```

### Q2. Few-shot prompt improvement (3 points)
Now improve the previous prompt by adding **two examples**.

Requirements:
- Store it in a variable called `few_shot_prompt`
- Include 2 input-output examples
- Ask the model to classify the 3 sample reviews after the examples

```python
few_shot_prompt = """
You are an AI system that analyzes customer opinions and determines their sentiment.

Sentiment categories:
- Positive
- Negative
- Neutral

Instructions:
- Read each review carefully
- Decide the overall sentiment
- Assign only one label per review

Examples:

Review: "The phone works perfectly and I absolutely love it."
Sentiment: Positive

Review: "The item stopped working after one day and support was useless."
Sentiment: Negative

Now classify the following reviews:

Output format:
Provide only the sentiment labels in a numbered list:
1. label
2. label
3. label

Reviews:
1. The product arrived late and the quality was poor.
2. Amazing battery life and elegant design.
3. It is fine for daily use, nothing special.
"""

print(few_shot_prompt)
```

## Reflection (optional, no extra points)
Briefly explain the difference between:
- sentiment analysis
- text classification
- prompt engineering

Sentiment analysis is basically about figuring out the feeling behind a text, like whether it’s positive, negative, or neutral. You usually see this in things like reviews or comments.

Text classification is more general. It means putting text into different categories based on what it’s about, like spam vs not spam, topic classification, or even sentiment. So sentiment analysis is actually one type of text classification.

Prompt engineering is about how we ask questions or give instructions to AI so it gives better answers. If we write the prompt clearly with proper instructions or examples, the output becomes much more accurate.
